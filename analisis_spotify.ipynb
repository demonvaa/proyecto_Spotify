```python
df = df.drop(columns=["Unnamed: 0"])
df.rename(columns= {'top genre':'genre'}, inplace= True)
```

```python
def generate_id_and_dim_table(name_column,DataFrame):
    lst_column = list(DataFrame[name_column].unique())
    new_df = pd.DataFrame(lst_column, columns=[name_column])
    new_id_column = DataFrame[name_column].map({name: i+1 for i,name in enumerate(lst_column)})
    return new_id_column, new_df
```

```python
df['artist'], df_artist = generate_id_and_dim_table('artist',df)
df['genre'], df_genre = generate_id_and_dim_table('genre',df)
df['country'], df_country = generate_id_and_dim_table('country',df)
```

```python
df.rename(columns= {'artist':'artist_id'}, inplace= True)
df.rename(columns= {'genre':'genre_id'}, inplace= True)
df.rename(columns= {'country':'country_id'}, inplace= True)
```

```python
from sqlalchemy import create_engine, text 
import config
```

```python
engine = create_engine(config.DATABASE_URL)
```

```python
# 2. Lectura del archivo con el schema
with open('schema_Spotify_proyect.sql', 'r', encoding='utf-8') as file:
    sql_schema = file.read()

# 3. Ejecución del schema en la base de datos
try:
    with engine.connect() as connection:
        # Dividimos por ";" por si hay varias instrucciones y las ejecutamos
        # Nota: Algunos drivers permiten ejecutar todo el bloque de golpe
        for statement in sql_schema.split(';'):
            if statement.strip():
                connection.execute(text(statement))
        connection.commit()
    print("Schema cargado correctamente.")
except Exception as e:
    print(f"Error al cargar el schema: {e}")
```

```python
# 1. Crear un motor que apunte DIRECTAMENTE a la base de datos
# Asegúrate de que la URL incluya el nombre de la DB al final
engine_db = create_engine(config.DATABASE_URL + "Spotify_proyect")

# 2. Volcado de las tablas usando el nuevo motor (engine_db)
try:
    df_country.to_sql('country', con=engine_db, if_exists='append', index=False)
    df_artist.to_sql('artist', con=engine_db, if_exists='append', index=False)
    df_genre.to_sql('genre', con=engine_db, if_exists='append', index=False)
    df.to_sql('top50_country', con=engine_db, if_exists='append', index=False)
    print("¡Datos cargados exitosamente en todas las tablas!")
except Exception as e:
    print(f"Error al cargar los datos: {e}")
```
